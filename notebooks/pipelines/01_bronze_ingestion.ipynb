```python
dbutils.widgets.text("run_id",         "")
dbutils.widgets.text("ingestion_date", "")
run_id         = dbutils.widgets.get("run_id")
ingestion_date = dbutils.widgets.get("ingestion_date")

print(f"run_id: {run_id} | ingestion_date: {ingestion_date}")
```

```python
args    = {"run_id": run_id, "ingestion_date": ingestion_date}
timeout = 3600

dbutils.notebook.run("../dml/bronze/01_play_history",    timeout, args)
dbutils.notebook.run("../dml/bronze/02_playlists",       timeout, args)
dbutils.notebook.run("../dml/bronze/03_playlist_tracks", timeout, args)
dbutils.notebook.run("../dml/bronze/04_tracks",          timeout, args)
dbutils.notebook.run("../dml/bronze/05_artists",         timeout, args)
dbutils.notebook.run("../dml/bronze/06_audio_features",  timeout, args)

print("Bronze ingestion complete.")
```
