# request_hash

Deterministic SHA-256 hash of `(method, url, params, body)` for deduplicating
API call log entries.

```python
import hashlib
import json as _json


def request_hash(method: str, url: str, params: dict, body: dict) -> str:
    payload = (
        f"{method.upper()}|"
        f"{url}|"
        f"{_json.dumps(params or {}, sort_keys=True)}|"
        f"{_json.dumps(body   or {}, sort_keys=True)}"
    )
    return hashlib.sha256(payload.encode()).hexdigest()
```
