# 03_dim_artists

DDL: `dim_artists` — Artist attributes with genre array.
`primary_genre` = `genres[0]` extracted by the DML transform.

```python
%run ../../tools/config/settings
```

```python
spark.sql(f"CREATE SCHEMA IF NOT EXISTS {CATALOG}.{SILVER_SCHEMA}")

spark.sql(f"""
CREATE TABLE IF NOT EXISTS {CATALOG}.{SILVER_SCHEMA}.dim_artists (
    artist_id           STRING      NOT NULL,
    artist_name         STRING,
    genres              ARRAY<STRING>,
    primary_genre       STRING      COMMENT 'genres[0], null if genres is empty',
    popularity          INT,
    followers_total     BIGINT,
    _valid_from         DATE
)
USING DELTA
COMMENT 'Silver dimension: artist attributes (SCD Type 1)'
TBLPROPERTIES (
    'delta.enableChangeDataFeed'       = 'true',
    'delta.columnMapping.mode'         = 'name'
)
""")

print(f"Table {CATALOG}.{SILVER_SCHEMA}.dim_artists ready.")
```
