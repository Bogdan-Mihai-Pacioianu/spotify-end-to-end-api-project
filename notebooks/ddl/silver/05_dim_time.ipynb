```python
CREATE TABLE IF NOT EXISTS spotify_etl.silver.dim_time (
  play_timestamp TIMESTAMP,
  hour_of_day INT,
  day_of_week INT,
  weekday_name STRING,
  is_weekend BOOLEAN,
  quarter STRING,
  year INT,
  month INT,
  day INT
)
USING DELTA
TBLPROPERTIES (
  delta.enableChangeDataFeed = true,
  delta.columnMapping.mode = 'name'
);
```
