# 05_dim_time

DDL: `dim_time` — Time attributes derived from `fct_plays.played_at`.
One row per unique play timestamp.

```python
%run ../../tools/config/settings
```

```python
spark.sql(f"CREATE SCHEMA IF NOT EXISTS {CATALOG}.{SILVER_SCHEMA}")

spark.sql(f"""
CREATE TABLE IF NOT EXISTS {CATALOG}.{SILVER_SCHEMA}.dim_time (
    played_at           TIMESTAMP   NOT NULL,
    play_date           DATE,
    play_hour           INT         COMMENT '0–23',
    day_of_week         INT         COMMENT '1=Sunday … 7=Saturday (Spark default)',
    day_name            STRING,
    week_of_year        INT,
    month               INT         COMMENT '1–12',
    month_name          STRING,
    year                INT,
    is_weekend          BOOLEAN     COMMENT 'True if day_of_week IN (1,7)'
)
USING DELTA
COMMENT 'Silver dimension: time attributes derived from play timestamps'
TBLPROPERTIES (
    'delta.columnMapping.mode' = 'name'
)
""")

print(f"Table {CATALOG}.{SILVER_SCHEMA}.dim_time ready.")
```
