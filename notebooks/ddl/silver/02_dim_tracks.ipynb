# 02_dim_tracks

DDL: `dim_tracks` — Normalized track attributes. `album_release_date` is cast
to DATE (Silver normalizes Spotify's partial-date strings).

```python
%run ../../tools/config/settings
```

```python
spark.sql(f"CREATE SCHEMA IF NOT EXISTS {CATALOG}.{SILVER_SCHEMA}")

spark.sql(f"""
CREATE TABLE IF NOT EXISTS {CATALOG}.{SILVER_SCHEMA}.dim_tracks (
    track_id            STRING      NOT NULL,
    track_name          STRING,
    duration_ms         BIGINT,
    duration_min        DOUBLE      COMMENT 'duration_ms / 60000',
    popularity          INT,
    explicit            BOOLEAN,
    album_id            STRING,
    album_name          STRING,
    album_release_date  DATE        COMMENT 'Parsed from Spotify partial-date string',
    _valid_from         DATE        COMMENT 'Ingestion date when record was first seen'
)
USING DELTA
COMMENT 'Silver dimension: normalized track attributes (SCD Type 1)'
TBLPROPERTIES (
    'delta.enableChangeDataFeed'       = 'true',
    'delta.columnMapping.mode'         = 'name'
)
""")

print(f"Table {CATALOG}.{SILVER_SCHEMA}.dim_tracks ready.")
```
