# 01_fct_plays

DDL: `fct_plays` — Deduplicated play events.
`play_id` = SHA256(`played_at` || `track_id`), used as the MERGE key.

```python
%run ../../tools/config/settings
```

```python
spark.sql(f"CREATE SCHEMA IF NOT EXISTS {CATALOG}.{SILVER_SCHEMA}")

spark.sql(f"""
CREATE TABLE IF NOT EXISTS {CATALOG}.{SILVER_SCHEMA}.fct_plays (
    play_id             STRING      NOT NULL COMMENT 'SHA256(played_at||track_id) — dedup key',
    played_at           TIMESTAMP   NOT NULL,
    track_id            STRING      NOT NULL,
    ingestion_date      DATE,
    run_id              STRING
)
USING DELTA
PARTITIONED BY (ingestion_date)
COMMENT 'Silver fact: deduplicated play events from bronze_play_history'
TBLPROPERTIES (
    'delta.enableChangeDataFeed'       = 'true',
    'delta.autoOptimize.optimizeWrite' = 'true',
    'delta.columnMapping.mode'         = 'name'
)
""")

print(f"Table {CATALOG}.{SILVER_SCHEMA}.fct_plays ready.")
```
