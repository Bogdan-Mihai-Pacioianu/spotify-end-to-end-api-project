```python
CREATE TABLE IF NOT EXISTS spotify_etl.bronze.bronze_playlist_tracks (
  playlist_id STRING,
  track_id STRING,
  track_name STRING,
  artist_ids ARRAY<STRING>,
  album_id STRING,
  added_at STRING,
  added_by STRING,
  duration_ms BIGINT,
  popularity INT
)
USING DELTA PARTITIONED BY (processing_date)
TBLPROPERTIES (
  delta.enableChangeDataFeed = true
);
```
