# 03_playlist_tracks

DDL: `bronze_playlist_tracks` — Raw track items from `/playlists/{id}/tracks`.
`playlist_id` is injected by the extraction pipeline.

```python
%run ../../tools/config/settings
```

```python
spark.sql(f"CREATE SCHEMA IF NOT EXISTS {CATALOG}.{BRONZE_SCHEMA}")

spark.sql(f"""
CREATE TABLE IF NOT EXISTS {CATALOG}.{BRONZE_SCHEMA}.bronze_playlist_tracks (
    playlist_id         STRING      COMMENT 'Parent playlist ID (injected by pipeline)',
    track_id            STRING,
    track_name          STRING,
    added_at            TIMESTAMP,
    added_by            STRING      COMMENT 'User ID who added the track',
    _raw                STRING,
    run_id              STRING,
    ingestion_date      DATE
)
USING DELTA
PARTITIONED BY (ingestion_date)
COMMENT 'Bronze: raw playlist tracks from Spotify /playlists/{{id}}/tracks'
TBLPROPERTIES (
    'delta.autoOptimize.optimizeWrite' = 'true',
    'delta.autoOptimize.autoCompact'   = 'true',
    'delta.columnMapping.mode'         = 'name'
)
""")

print(f"Table {CATALOG}.{BRONZE_SCHEMA}.bronze_playlist_tracks ready.")
```
