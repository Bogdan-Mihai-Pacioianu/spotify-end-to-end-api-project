```python
catalog = "spotify_etl"
layer = "bronze"
name_table = "bronze_playlist_tracks"
```

```python
spark.sql(f"""
              CREATE TABLE IF NOT EXISTS {catalog}.{layer}.{name_table} (
                  playlist_id STRING,
                  track_id STRING,
                  track_name STRING,
                  artist_ids ARRAY<STRING>,
                  album_id STRING,
                  added_at STRING,
                  added_by STRING,
                  duration_ms BIGINT,
                  popularity INT,
                  processing_date DATE
              )
              USING DELTA PARTITIONED BY (processing_date)
              TBLPROPERTIES (
                delta.enableChangeDataFeed = true
              )
""")
```
