```python
catalog = "spotify_etl"
layer = "bronze"
name_table = "bronze_artists"
```

```python
spark.sql(f"""
          CREATE TABLE IF NOT EXISTS {catalog}.{layer}.{name_table} (
            artist_id STRING,
            artist_name STRING,
            genres ARRAY<STRING>,
            followers BIGINT,
            popularity INT,
            uri STRING,
            processing_date DATE
          )
          USING DELTA PARTITIONED BY (processing_date)
          TBLPROPERTIES (
            delta.enableChangeDataFeed = true
          )
""")
```
