```python
catalog = "spotify_etl"
layer = "bronze"
name_table = "bronze_tracks"
```

```python
spark.sql(f"""
          CREATE TABLE IF NOT EXISTS {catalog}.{layer}.{name_table} (
              track_id STRING,
              track_name STRING,
              album_id STRING,
              album_name STRING,
              artist_id STRING,
              artist_name STRING,
              duration_ms BIGINT,
              popularity INT,
              explicit BOOLEAN,
              release_date STRING,
              preview_url STRING,
              processing_date DATE
            )
            USING DELTA PARTITIONED BY (processing_date)
            TBLPROPERTIES (
            delta.enableChangeDataFeed = true
            )
""")
```
