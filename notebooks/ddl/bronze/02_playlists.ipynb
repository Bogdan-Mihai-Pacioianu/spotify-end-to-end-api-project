```python
catalog = "spotify_etl"
layer = "bronze"
name_table = "bronze_playlists"
```

```python
spark.sql(f"""
          CREATE TABLE IF NOT EXISTS {catalog}.{layer}.{name_table} (
            playlist_id STRING,
            playlist_name STRING,
            owner_name STRING,
            followers BIGINT,
            total_tracks BIGINT,
            description STRING,
            snapshot_id STRING,
            processing_date DATE
          )
          USING DELTA PARTITIONED BY (processing_date)
          TBLPROPERTIES (
            delta.enableChangeDataFeed = true
          )
""")
```
