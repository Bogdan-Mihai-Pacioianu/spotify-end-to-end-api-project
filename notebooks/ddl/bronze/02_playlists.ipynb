```python
CREATE TABLE IF NOT EXISTS spotify_etl.bronze.bronze_playlists (
  playlist_id STRING,
  playlist_name STRING,
  owner_name STRING,
  followers BIGINT,
  total_tracks BIGINT,
  description STRING,
  snapshot_id STRING
)
USING DELTA PARTITIONED BY (processing_date)
TBLPROPERTIES (
  delta.enableChangeDataFeed = true
);
```
