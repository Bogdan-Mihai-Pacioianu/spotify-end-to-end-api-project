# 02_top_tracks

DDL: `gold_top_tracks` — Top tracks ranked by play count per period (`7d`, `30d`, `all`).

```python
%run ../../tools/config/settings
```

```python
spark.sql(f"CREATE SCHEMA IF NOT EXISTS {CATALOG}.{GOLD_SCHEMA}")

spark.sql(f"""
CREATE TABLE IF NOT EXISTS {CATALOG}.{GOLD_SCHEMA}.gold_top_tracks (
    snapshot_date       DATE        NOT NULL,
    period              STRING      NOT NULL    COMMENT '7d / 30d / all',
    rank                INT         NOT NULL,
    track_id            STRING,
    track_name          STRING,
    play_count          BIGINT
)
USING DELTA
PARTITIONED BY (snapshot_date)
COMMENT 'Gold: top tracks ranked by play count per period'
TBLPROPERTIES (
    'delta.columnMapping.mode' = 'name'
)
""")

print(f"Table {CATALOG}.{GOLD_SCHEMA}.gold_top_tracks ready.")
```
