# 01_user_listening_summary

DDL: `gold_user_listening_summary` — Cumulative listening totals per snapshot date.
One row per daily pipeline run.

```python
%run ../../tools/config/settings
```

```python
spark.sql(f"CREATE SCHEMA IF NOT EXISTS {CATALOG}.{GOLD_SCHEMA}")

spark.sql(f"""
CREATE TABLE IF NOT EXISTS {CATALOG}.{GOLD_SCHEMA}.gold_user_listening_summary (
    snapshot_date       DATE        NOT NULL,
    total_plays         BIGINT,
    unique_tracks       BIGINT,
    unique_artists      BIGINT,
    total_minutes       DOUBLE
)
USING DELTA
COMMENT 'Gold: cumulative listening summary per snapshot date'
TBLPROPERTIES (
    'delta.columnMapping.mode' = 'name'
)
""")

print(f"Table {CATALOG}.{GOLD_SCHEMA}.gold_user_listening_summary ready.")
```
