```python
def chunk_list(data, size):
    """Splits a list into smaller chunks."""
    for i in range(0, len(data), size):
        yield data[i:i + size]
```

```python
import os
import json
import re

def _find_all_raw_files(entity_path):
    """Returns dicts mapping page_idx -> (page_data, meta_data) for all JSON files."""
    pages = {}  # page_index -> {"data": ..., "meta": ...}
    meta_pages = {}  # For playlist_tracks: (page_idx_local) -> playlist_id mapping
    results = []
    
    for root, dirs, files in os.walk(entity_path):
        for file in files:
            full_path = os.path.join(root, file)
            if file.startswith("page_") and file.endswith("_meta.json"):
                m = re.match(r'page_(\d+)_meta\.json', file)
                if m:
                    idx = int(m.group(1))
                    try:
                        with open(full_path, 'r', encoding='utf-8') as f:
                            meta_pages[idx] = json.load(f)
                    except (json.JSONDecodeError, IOError):
                        pass
            elif file.startswith("page_") and file.endswith(".json"):
                m = re.match(r'page_(\d+)\.json', file)
                if m:
                    idx = int(m.group(1))
                    try:
                        with open(full_path, 'r', encoding='utf-8') as f:
                            data = json.load(f)
                            data["_meta_page_idx"] = idx
                            results.append(data)
                    except (json.JSONDecodeError, IOError) as e:
                        print(f"Warning: could not parse {file}: {e}")
    return results, meta_pages

def get_play_history(unique_track_ids, unique_artist_ids):
    print("Fetching play history from raw files...")
    play_history_data = []
    entity_path = f"{RAW_BASE_PATH}/recently_played"
    
    pages, _ = _find_all_raw_files(entity_path)
    for page in pages:
        for item in page.get('items', []):
            if not item:
                continue
            track = item.get('track', {})
            artists = track.get('artists') or []
            artist = artists[0] if artists else {}
            
            play_history_data.append({
                'played_at': item.get('played_at'),
                'track_id': track.get('id'),
                'track_name': track.get('name'),
                'artist_id': artist.get('id'),
                'artist_name': artist.get('name'),
                'album_id': track.get('album', {}).get('id'),
                'album_name': track.get('album', {}).get('name'),
                'context_type': (item.get('context') or {}).get('type'),
                'duration_ms': track.get('duration_ms')
            })
            if track.get('id'): unique_track_ids.add(track.get('id'))
            if artist.get('id'): unique_artist_ids.add(artist.get('id'))
            
    print(f"✅ Retrieved {len(play_history_data)} play history records.")
    return play_history_data

def get_playlists_and_tracks(unique_track_ids, unique_artist_ids):
    print("Fetching playlists and their tracks from raw files...")
    playlists_data = []
    playlist_tracks_data = []
    
    # Extract Playlists
    pages_p, _ = _find_all_raw_files(f"{RAW_BASE_PATH}/me_playlists")
    for page in pages_p:
        for playlist in page.get('items', []):
            playlist_id = playlist.get('id')
            if not playlist_id:
                continue
            
            followers_total = None
            followers_obj = playlist.get('followers')
            if isinstance(followers_obj, dict):
                followers_total = followers_obj.get('total')
                
            playlists_data.append({
                'playlist_id': playlist_id,
                'playlist_name': playlist.get('name'),
                'owner_name': (playlist.get('owner') or {}).get('display_name'),
                'followers': followers_total,
                'total_tracks': (playlist.get('tracks') or {}).get('total'),
                'description': playlist.get('description'),
                'snapshot_id': playlist.get('snapshot_id')
            })
            
    # Extract Playlist Tracks
    pages_t, meta_map = _find_all_raw_files(f"{RAW_BASE_PATH}/playlist_tracks")
    
    # Build playlist_id lookup from meta files using page_idx_local
    pid_by_page = {}
    for meta_page_idx, meta in meta_map.items():
        pid = meta.get('playlist_id')
        local_idx = meta.get('page_idx_local')
        if pid and local_idx is not None:
            # Use local_idx as the key since it's the sequential page counter in ingest_playlist_tracks
            pid_by_page[local_idx] = pid

    # For playlist_tracks, pages are numbered globally (page_idx_global in notebook 00)
    # but meta files have page_idx_local and playlist_id
    # We need to map each page file to its playlist_id
    # The meta files are saved alongside page files with matching numeric suffix
    meta_playlist_mapping = {}
    for idx, m in meta_map.items():
        pid = m.get('playlist_id')
        if pid:
            meta_playlist_mapping[idx] = pid
    
    for page in pages_t:
        page_idx = page.get('_meta_page_idx')
        
        # Try meta mapping first
        page_pid = meta_playlist_mapping.get(page_idx)
        
        # Fallback: regex from href
        if page_pid is None:
            href = page.get('href', '')
            m = re.search(r'playlists/([^/]+)/tracks', href)
            page_pid = m.group(1) if m else None
        
        if page_pid is None:
            print(f"Warning: could not extract playlist_id for page {page_idx}, skipping")
            continue
        
        for item in page.get('items', []):
            track = item.get('track')
            if not track or not track.get('id'):
                continue
                
            track_artist_ids = [a.get('id') for a in (track.get('artists') or []) if a.get('id')]
            
            playlist_tracks_data.append({
                'playlist_id': page_pid,
                'track_id': track.get('id'),
                'track_name': track.get('name'),
                'artist_ids': track_artist_ids,
                'album_id': (track.get('album') or {}).get('id'),
                'added_at': item.get('added_at'),
                'added_by': (item.get('added_by') or {}).get('id'),
                'duration_ms': track.get('duration_ms'),
                'popularity': track.get('popularity')
            })
            
            unique_track_ids.add(track.get('id'))
            for art_id in track_artist_ids:
                unique_artist_ids.add(art_id)
                
    print(f"✅ Retrieved {len(playlists_data)} playlists and {len(playlist_tracks_data)} playlist-track mappings.")
    return playlists_data, playlist_tracks_data
    
def get_full_metadata(unique_track_ids, unique_artist_ids):
    print(f"Fetching metadata for tracks and artists from raw files...")
    tracks_data, artists_data = [], []
    
    pages_tr, _ = _find_all_raw_files(f"{RAW_BASE_PATH}/tracks_bulk")
    for page in pages_tr:
        for track in page.get('tracks', []):
            if not track:
                continue
            artists = track.get('artists') or []
            first_artist = artists[0] if artists else {}
            album = track.get('album') or {}
            tracks_data.append({
                'track_id': track.get('id'),
                'track_name': track.get('name'),
                'album_id': album.get('id'),
                'album_name': album.get('name'),
                'artist_id': first_artist.get('id'),
                'artist_name': first_artist.get('name'),
                'duration_ms': track.get('duration_ms'),
                'popularity': track.get('popularity'),
                'explicit': track.get('explicit'),
                'release_date': album.get('release_date'),
                'preview_url': track.get('preview_url')
            })
            
    pages_ar, _ = _find_all_raw_files(f"{RAW_BASE_PATH}/artists_bulk")
    for page in pages_ar:
        for artist in page.get('artists', []):
            if not artist:
                continue
            followers_total = None
            followers_obj = artist.get('followers')
            if isinstance(followers_obj, dict):
                followers_total = followers_obj.get('total')
                
            artists_data.append({
                'artist_id': artist.get('id'),
                'artist_name': artist.get('name'),
                'genres': artist.get('genres'),
                'followers': followers_total,
                'popularity': artist.get('popularity'),
                'uri': artist.get('uri')
            })
            
    print(f"✅ Retrieved: {len(tracks_data)} tracks, {len(artists_data)} artists.")
    return tracks_data, artists_data

def get_audio_features():
    print("Fetching audio features from raw files...")
    audio_features_data = []
    
    pages, _ = _find_all_raw_files(f"{RAW_BASE_PATH}/audio_features_bulk")
    for page in pages:
        for feature in page.get('audio_features', []):
            if not feature:
                continue
            audio_features_data.append({
                'track_id': feature.get('id'),
                'danceability': feature.get('danceability'),
                'energy': feature.get('energy'),
                'key': feature.get('key'),
                'loudness': feature.get('loudness'),
                'mode': feature.get('mode'),
                'speechiness': feature.get('speechiness'),
                'acousticness': feature.get('acousticness'),
                'instrumentalness': feature.get('instrumentalness'),
                'liveness': feature.get('liveness'),
                'valence': feature.get('valence'),
                'tempo': feature.get('tempo'),
                'duration_ms': feature.get('duration_ms'),
                'time_signature': feature.get('time_signature')
            })
    
    print(f"✅ Retrieved {len(audio_features_data)} audio feature records.")
    return audio_features_data
```

```python
def save_as_delta_table(all_data):
    """
    Saves each list of dictionaries as a Delta table
    in the 'bronze' schema of Unity Catalog.
    """
    print(f"\n--- STARTING DATA LOAD (Saving to Delta Lake in {CATALOG_NAME}.{BRONZE_SCHEMA}) ---")
    
    for table_name, data_list in all_data.items():
        if not data_list:
            print(f"Skipping '{table_name}', no data found.")
            continue
        
        try:
            # Create Spark DataFrame directly from list of dicts (no pandas intermediary)
            df = spark.createDataFrame(data_list)
            from pyspark.sql import functions as F
            df = df.withColumn("processing_date", F.current_date())
            
            full_table_name = f"{CATALOG_NAME}.{BRONZE_SCHEMA}.{table_name}"
            print(f"Writing {df.count()} rows to {full_table_name}...")
            df.write.format("delta").mode("append").partitionBy("processing_date").option("mergeSchema", "true").saveAsTable(full_table_name)
            
            print(f"Successfully saved: {full_table_name}")
            
        except Exception as e:
            print(f"ERROR saving {table_name}: {e}")
            raise  # Re-raise so failures don't go silently
```

```python
print("--- STARTING SPOTIFY BRONZE PIPELINE ---")

unique_track_ids = set()
unique_artist_ids = set()

# --- 1. EXTRACT ---
print("\n--- STEP 1: EXTRACTION (Raw files -> Python) ---")
play_history_table = get_play_history(unique_track_ids, unique_artist_ids)
playlists_table, playlist_tracks_table = get_playlists_and_tracks(unique_track_ids, unique_artist_ids)

# --- 2. ENRICH ---
print("\n--- STEP 2: ENRICHMENT (Raw files -> Python) ---")
tracks_table, artists_table = get_full_metadata(unique_track_ids, unique_artist_ids)
audio_features_table = get_audio_features()

print("\n--- EXTRACTION SUMMARY ---")
print(f"bronze_play_history data:     {len(play_history_table)} rows")
print("------------------------------")
print(f"bronze_playlists data:        {len(playlists_table)} rows")
print(f"bronze_playlist_tracks data:  {len(playlist_tracks_table)} rows")
print("------------------------------")
print(f"bronze_tracks data:           {len(tracks_table)} rows")
print(f"bronze_artists data:          {len(artists_table)} rows")
print(f"bronze_audio_features data:   {len(audio_features_table)} rows")

# --- 3. LOAD (Python -> Delta) ---
print("\n--- STEP 3: LOAD (Python -> Delta) ---")
all_bronze_data = {
    "bronze_play_history": play_history_table,
    "bronze_playlists": playlists_table,
    "bronze_playlist_tracks": playlist_tracks_table,
    "bronze_tracks": tracks_table,
    "bronze_artists": artists_table,
    "bronze_audio_features": audio_features_table,
}

save_as_delta_table(all_bronze_data)

print("\n--- BRONZE PIPELINE FINISHED ---")
```
